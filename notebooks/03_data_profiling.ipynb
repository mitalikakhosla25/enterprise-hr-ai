```python
import pandas as pd
from pathlib import Path

DATA_PROCESSED = Path("../data/processed")

hr = pd.read_csv(DATA_PROCESSED / "hr_cleaned.csv")

print("=" * 80)
print("PROCESSED HR DATA LOADED")
print("=" * 80)

print("Shape:", hr.shape)
print("Unique Employees:", hr["Employee ID"].nunique())
print("Duplicate IDs:", hr["Employee ID"].duplicated().sum())

hr.head()

hr.dtypes
```

```python
print("=" * 80)
print("DATASET OVERVIEW")
print("=" * 80)

print(f"Rows                  : {len(hr):,}")
print(f"Columns               : {hr.shape[1]}")
print(f"Unique Employees      : {hr['Employee ID'].nunique():,}")
print(f"Duplicate Employee IDs: {hr['Employee ID'].duplicated().sum():,}")
print(f"Total Missing Cells   : {hr.isna().sum().sum():,}")

print("\nColumn Groups")
print("-" * 80)

numeric_cols = hr.select_dtypes(include="number").columns.tolist()
categorical_cols = hr.select_dtypes(include="object").columns.tolist()
boolean_cols = hr.select_dtypes(include="bool").columns.tolist()

print(f"Numeric columns     : {len(numeric_cols)}")
print(f"Categorical columns : {len(categorical_cols)}")
print(f"Boolean columns     : {len(boolean_cols)}")
```

```python
print("\nNUMERIC COLUMNS")
print("-" * 80)

print(numeric_cols)

print("\nCATEGORICAL COLUMNS")
print("-" * 80)

print(categorical_cols)

print("\nBOOLEAN COLUMNS")
print("-" * 80)

print(boolean_cols)
```

```python
missing_profile = (
    hr.isna()
      .sum()
      .to_frame("MissingCount")
)

missing_profile["MissingPercentage"] = (
    missing_profile["MissingCount"] / len(hr) * 100
)

missing_profile = (
    missing_profile[missing_profile["MissingCount"] > 0]
    .sort_values("MissingCount", ascending=False)
)

print("=" * 80)
print("MISSINGNESS PROFILE")
print("=" * 80)

print(missing_profile)
```

```python
print("=" * 80)
print("EMPLOYEE STATUS PROFILE")
print("=" * 80)

status_profile = (
    hr["EmployeeStatus"]
    .value_counts(dropna=False)
    .to_frame("EmployeeCount")
)

status_profile["Percentage"] = (
    status_profile["EmployeeCount"] / len(hr) * 100
)

print(status_profile)
```

```python
print("\nEMPLOYEE TYPE")
print("-" * 80)
print(hr["EmployeeType"].value_counts())

print("\nDEPARTMENT")
print("-" * 80)
print(hr["DepartmentType"].value_counts())

print("\nBUSINESS UNIT")
print("-" * 80)
print(hr["BusinessUnit"].value_counts())
```

```python
score_columns = [
    "Current Employee Rating",
    "Engagement Score",
    "Satisfaction Score",
    "Work-Life Balance Score",
    "Performance Score"
]

for col in score_columns:
    print("\n" + "=" * 80)
    print(col)
    print("=" * 80)
    print(hr[col].value_counts(dropna=False).sort_index())
```

```python
print("=" * 80)
print("DATA QUALITY FLAGS")
print("=" * 80)

flag_columns = [
    "DOBMissing",
    "TrainingBeforeStart",
    "SurveyBeforeStart"
]

for col in flag_columns:
    count = hr[col].sum()
    percentage = count / len(hr) * 100

    print(
        f"{col:25} : "
        f"{count:4} employees "
        f"({percentage:.2f}%)"
    )
```

```python
print("=" * 80)
print("TENURE PROFILE")
print("=" * 80)

# Convert dates temporarily for profiling
hr["StartDate"] = pd.to_datetime(hr["StartDate"], errors="coerce")
hr["ExitDate"] = pd.to_datetime(hr["ExitDate"], errors="coerce")

# Use the latest date available in the HR dataset as reference
reference_date = max(
    hr["StartDate"].max(),
    hr["Survey Date"].pipe(pd.to_datetime).max()
)

hr["TenureYears"] = (
    reference_date - hr["StartDate"]
).dt.days / 365.25

print("Reference date:", reference_date)
print("\nTenure statistics:")
print(hr["TenureYears"].describe())
```

```python
print("\nTENURE GROUPS")
print("-" * 80)

tenure_groups = pd.cut(
    hr["TenureYears"],
    bins=[-1, 1, 3, 5, 10, float("inf")],
    labels=[
        "0-1 years",
        "1-3 years",
        "3-5 years",
        "5-10 years",
        "10+ years"
    ]
)

print(tenure_groups.value_counts().sort_index())
```

```python
print("=" * 80)
print("TRAINING PROFILE")
print("=" * 80)

print("Training Programs:")
print(hr["Training Program Name"].value_counts())

print("\nTraining Types:")
print(hr["Training Type"].value_counts())

print("\nTraining Outcomes:")
print(hr["Training Outcome"].value_counts())

print("\nTraining Duration:")
print(hr["Training Duration(Days)"].describe())

print("\nTraining Cost:")
print(hr["Training Cost"].describe())
```

```python
print("=" * 80)
print("TRAINING OUTCOME × PERFORMANCE")
print("=" * 80)

training_performance = pd.crosstab(
    hr["Training Outcome"],
    hr["Performance Score"],
    normalize="index"
) * 100

print(training_performance.round(2))
```

```python
print("=" * 80)
print("PERFORMANCE BY DEPARTMENT")
print("=" * 80)

performance_department = pd.crosstab(
    hr["DepartmentType"],
    hr["Performance Score"],
    normalize="index"
) * 100

print(performance_department.round(2))
```

```python
print("=" * 80)
print("AVERAGE HR SCORES BY DEPARTMENT")
print("=" * 80)

score_columns = [
    "Current Employee Rating",
    "Engagement Score",
    "Satisfaction Score",
    "Work-Life Balance Score"
]

department_scores = (
    hr.groupby("DepartmentType")[score_columns]
      .mean()
      .round(2)
)

print(department_scores)
```

```python
from pathlib import Path
import pandas as pd

DATA_RAW = Path("../data/raw")
DATA_PROCESSED = Path("../data/processed")

# Main processed HR dataset
hr = pd.read_csv(
    DATA_PROCESSED / "hr_cleaned.csv"
)

# Attrition dataset
attrition_hr = pd.read_csv(
    DATA_RAW / "WA_Fn-UseC_-HR-Employee-Attrition.csv"
)

print("=" * 80)
print("DATASETS LOADED FOR PROFILING")
print("=" * 80)

print("HR dataset        :", hr.shape)
print("Attrition dataset :", attrition_hr.shape)
```

```python
print("=" * 80)
print("ATTRITION DATASET PROFILE")
print("=" * 80)

print("Shape:", attrition_hr.shape)
print("Unique Employee Numbers:", attrition_hr["EmployeeNumber"].nunique())

print("\nAttrition Distribution")
print("-" * 80)

attrition_counts = attrition_hr["Attrition"].value_counts()

print(attrition_counts)

print("\nAttrition Percentage")
print("-" * 80)

print(
    (attrition_counts / len(attrition_hr) * 100).round(2)
)
```

```python
print("=" * 80)
print("NUMERICAL FEATURES VS ATTRITION")
print("=" * 80)

numeric_attrition = [
    "Age",
    "DailyRate",
    "DistanceFromHome",
    "HourlyRate",
    "JobLevel",
    "MonthlyIncome",
    "MonthlyRate",
    "NumCompaniesWorked",
    "PercentSalaryHike",
    "PerformanceRating",
    "StockOptionLevel",
    "TotalWorkingYears",
    "TrainingTimesLastYear",
    "WorkLifeBalance",
    "YearsAtCompany",
    "YearsInCurrentRole",
    "YearsSinceLastPromotion",
    "YearsWithCurrManager"
]

attrition_numeric_profile = (
    attrition_hr
    .groupby("Attrition")[numeric_attrition]
    .mean()
    .T
    .round(2)
)

print(attrition_numeric_profile)
```

```python
categorical_attrition = [
    "BusinessTravel",
    "Department",
    "EducationField",
    "Gender",
    "JobRole",
    "MaritalStatus",
    "OverTime"
]

for col in categorical_attrition:

    print("\n" + "=" * 80)
    print(f"{col.upper()} VS ATTRITION")
    print("=" * 80)

    profile = (
        pd.crosstab(
            attrition_hr[col],
            attrition_hr["Attrition"],
            normalize="index"
        ) * 100
    )

    print(profile.round(2))
```
